### 2.1 The protocol, fixed before looking at fairness

* **Unit**: a (candidate, job) pair.
* **Selected**: a candidate is selected for a job if they are in the top half of that job's candidates, ranked by
  score. A score means something only in comparison with the other candidates of the same job, so the ranking is done
  inside each job, as in a direct matching. As a stricter second outcome we also use the top 3.
* **Ties**: when candidates of the same job have the same score, their order is drawn at random. Each pair gets one
  random number, and the same number is used with every score (Adecco, P, O, C). So every job with $n$ candidates selects exactly
  $\lceil n/2 \rceil$ of them, and a pair can change outcome between two versions only because the scores change, never because of a
  different draw.
* **Jobs kept**: jobs with at least 5 candidates. In a shorter list, being in the top half says nothing. Candidates
  with gender `Other` or `Unknown` (117) are left out: they are too few for group statistics.
* **Scores**: Adecco's own score (`score_adecco`) and the score recomputed with our formula on the three profile
  versions (`score_P`, `score_O`, `score_C`).
* **Groups**: `Gender`, `Location`, and the two together (`group`, e.g. `Female_SOUTH`).
* **Confidence intervals** (Phase 3): we resample candidates, not pairs, because the same candidate appears in several
  pairs.

The file's `rev_match_rank` is the position of the job among the candidate's recommendations, so we do not use it:
the position among the job's candidates is computed from the score. The file holds only the jobs recommended to each
candidate, so a job's list contains the candidates it was recommended to, not every candidate who applied. The
ranking is among those.

```python
from hiring_cv_bias.fairness.protocol import (
    SCORES,
    add_outcomes,
    filter_protocol,
    load_pairs,
    protocol_sizes,
)

pairs = load_pairs()
df = add_outcomes(filter_protocol(pairs), SCORES)
n_jobs, n_cand = pairs["Job_id"].n_unique(), pairs["CANDIDATE_ID"].n_unique()
print("kept by the protocol (jobs with at least 5 Female or Male candidates):")
print(
    f"  pairs:      {df.height:,} of {pairs.height:,} ({df.height / pairs.height:.1%})"
)
print(f"  jobs:       {df['Job_id'].n_unique():,} of {n_jobs:,}")
print(f"  candidates: {df['CANDIDATE_ID'].n_unique():,} of {n_cand:,}")
print(
    f"median number of candidates per job: {df.group_by('Job_id').len()['len'].median():.0f}"
)
print(
    f"median number of jobs per candidate: {df.group_by('CANDIDATE_ID').len()['len'].median():.0f}"
)
display(protocol_sizes(df, "group"))
```

The filter keeps three quarters of the pairs, in lists with a median of nine candidates. The North holds most of the
data; the other four groups are much smaller, with a few hundred candidates and about two thousand pairs each. A list
mixes candidates of different groups, so a selection is a competition between groups inside the same job. The `jobs`
column counts, for each group, the lists in which it has at least one candidate.

### 2.2 The outcomes under the four scores

We rank the candidates of each job four times: with Adecco's score (`adecco`) and with our recomputed score on P, O
and C. Each row of the table is one of these rankings:

* `decided_by_draw`: share of pairs whose outcome is decided by the random draw, because they have the same score as
  another candidate of the job and the two sit on the two sides of the cutoff;
* `agree_adecco`: share of pairs that get the same outcome (selected or not) as with Adecco's score.

The first table uses `top_half` as the outcome, the second `top3`. Below the tables: the share of selected pairs, and
the pairs whose outcome changes when the parser is corrected (O → C).

```python
from hiring_cv_bias.fairness.protocol import VERSIONS, outcome_summary

display(outcome_summary(df, VERSIONS, "top_half"))
display(outcome_summary(df, VERSIONS, "top3"))
print(
    f"selected pairs (top_half), the same for every score: {df['top_half_O'].mean():.1%}"
)
gain = (~df["top_half_O"] & df["top_half_C"]).sum()
loss = (df["top_half_O"] & ~df["top_half_C"]).sum()
print(
    f"pairs whose top_half outcome changes from O to C: {(gain + loss) / df.height:.1%} "
    f"({gain:,} enter the top half, {loss:,} leave it)"
)
```

What the tables say:

* **Many of Adecco's decisions are a tie.** With Adecco's score, about one pair in five has the same score as another
  candidate exactly at the cutoff of the top half, so the draw decides its outcome. With P almost as many; with O and
  C much fewer.
* **A little more than half of the pairs are selected, with every score.** The draw makes every job select exactly
  $\lceil n/2 \rceil$ candidates. It is above 50% because lists with an odd number of candidates round up
  ($\lceil 5/2 \rceil = 3$ selected).
* **Our score reproduces Adecco's decision.** P agrees with Adecco on almost every pair, with both outcomes. This
  validates the reconstruction: formula, rules and estimate together.
* **The parser changes the decision of about one pair in five.** This is the distance between O and Adecco.
* **Correcting the parser changes less than one decision in ten.** As many pairs enter the top half as leave it: the
  number of places in a job is fixed, so when a candidate enters, another one leaves. C agrees with Adecco more often
  than O. Phase 3 asks who these pairs are: whether some groups enter or leave more often than others.
