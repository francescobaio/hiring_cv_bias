# Do CV-parser errors reach the selection? A fairness audit of the Adecco reverse matching

**AI Ethics, Module 2 – Hiring CV Bias** · Francesco Baiocchi, Leonardo Petrilli

**Where this study comes from.** In the Hiring CV Bias project we studied a proprietary parser. The parser reads
anonymized CVs and extracts skills from them for Adecco. We cleaned the CVs and the parser output (`data_cleaning`),
we looked for imbalances between demographic groups in the extracted skills (`distributions_analysis`), and we
measured the parser's errors directly (`bias_detection`). For that last step we wrote our own extractors: rule-based
ones for driving licenses and language skills, and exact and semantic matching for job titles. They found information
that is in the CV text but missing from the parser output, and we compared the miss rates across gender and
macro-region. Those false negatives are the starting point here.

**The question.** A parser error matters only if it changes a decision. Adecco's *reverse matching* computes a match
score for every (candidate, job) pair, from the candidate's profile and the job's requirements, and recommends to each
candidate their 10 best jobs. The same scores, read from the job's side, give the *direct matching* that a recruiter
uses: a job and its candidates, ordered by score. Our data is the reverse matching, and we read it as a direct
matching: for each job, the candidates whose top 10 contains it, ranked by score. This study asks whether the parser's
errors reach that ranking, and whether they hit some groups more than others.

* **RQ1** – Do the errors of the CV parser propagate to the selection on the job offers, and do they penalize some
  groups (gender, macro-region, women of the South) more than others?
* **RQ2** – Does correcting the parser work as a fairness mitigation, compared with the standard techniques
  (reweighing, in-processing, post-processing)?
* **RQ3** – Which pieces of information drive the effect, and how robust are the conclusions to our choices?

**Pipeline under study**

```
CV ──parser──► profile ──matcher──► score for every (candidate, job) ──► position among the job's candidates
```

**Three versions of every candidate profile**

| version | content | role |
|---|---|---|
| **P** | the real Adecco profile (`Candidates` sheet) | validates the reconstructed matcher |
| **O** – Original | licenses, languages, job titles **only as read by the parser** | the fully automatic pipeline |
| **C** – Corrected | O + the false negatives recovered from the CV text | the pre-processing mitigation |

### 1.2 The score: outliers and the formula

For every pair, `ReverseMatching.xlsx` gives the `match_score` and the seven components computed by Adecco, but not
how they are combined.

#### Range of the components

For each column the table gives the number of distinct values and the range. Four components are binary (0 or 1):
`Interest`, `Card_match`, `Languages_match` and `Regulated_Trainings_match`. `Distance_match` takes three values
(every pair is within 25 km). `match_skills` (a share) and `Experience` (a similarity) are continuous. So every
component is in [0, 1], while `match_score` covers a much wider range. We look at its distribution next.

````python
from hiring_cv_bias.reverse_matching.scoring import COMPONENTS

display(
    pairs_all.select(
        pl.struct(
            n_values=pl.col(c).n_unique(), min=pl.col(c).min(), max=pl.col(c).max()
        ).alias(c)
        for c in COMPONENTS + ["match_score"]
    )
    .unpivot(variable_name="column")
    .unnest("value")
)
````

#### Distance bands

`Distance_match` takes only three values, but `ReverseMatching.xlsx` also gives the distance in km (`distance_km`). If
we group the pairs by `Distance_match`, we see the km range behind each value: 1 below 10 km, 0.6 from 10 to 20 km,
0.2 from 20 to 25 km. The ranges do not overlap, so `Distance_match` is a step function of the distance. We use these
bands to recompute it in Component rules.

````python
display(
    pairs_all.group_by("Distance_match")
    .agg(
        pl.len().alias("pairs"),
        pl.col("distance_km").min().alias("km_min"),
        pl.col("distance_km").max().alias("km_max"),
    )
    .sort("Distance_match", descending=True)
)
````

#### Outliers

The box is narrow and most scores are low. A small group of pairs forms a long right tail, up to several times the
median. We call them outliers, with the standard box-plot rule: a pair is an outlier when its score is above the upper
whisker (red dashed line: third quartile + 1.5 × box height). The lower whisker is below the minimum score, so there
are no low outliers. The second printed line counts them: about one pair in fifty, concentrated in a small number of
jobs.

The distribution alone cannot tell us whether the outliers follow the same rule as the other pairs. So we study how
the score is built on the other pairs, and then we test the outliers against the result (Exact check). We set aside
the pairs only, not the candidates.

````python
from hiring_cv_bias.fairness.plots import plot_score_distribution
from hiring_cv_bias.reverse_matching.scoring import score_outliers

q1, q3 = pairs_all.select(
    pl.col("match_score").quantile(0.25).alias("q1"),
    pl.col("match_score").quantile(0.75).alias("q3"),
).row(0)
threshold = q3 + 1.5 * (q3 - q1)
is_outlier = score_outliers(pairs_all)
pairs_clean = pairs_all.filter(~is_outlier)

fig = plot_score_distribution(
    pairs_all["match_score"].to_list(),
    threshold,
    title=f"match_score of the {pairs_all.height:,} pairs",
)
plt.show()
````

````python
outliers = pairs_all.filter(is_outlier)
print(
    f"box: Q1 = {q1:.2f}, Q3 = {q3:.2f} | lower whisker = {q1 - 1.5 * (q3 - q1):.2f}, "
    f"minimum score = {pairs_all['match_score'].min():.2f}"
)
print(
    f"outliers: {outliers.height} pairs ({outliers.height / pairs_all.height:.1%}) above {threshold:.2f}, "
    f"in {outliers['Job_id'].n_unique()} of {pairs_all['Job_id'].n_unique()} jobs"
)
````

#### Weights

We start from the simplest hypothesis, a weighted sum, fitted by least squares on the pairs without outliers:

$$\text{match\_score} \approx w_0 + w_1\,\text{Experience} + w_2\,\text{Interest} + \dots + w_7\,\text{Distance\_match}$$

The estimates are almost round numbers. Rounded to one decimal:

$$\text{score} = 2\,\text{Experience} + \text{Interest} + \text{match\_skills} + \text{Distance} + 0.3\,(\text{Card} + \text{Languages} + \text{Regulated})$$

The last row is the same regression on all pairs, outliers included. There the estimates are not round any more:
`Experience` weighs three times as much, one weight becomes negative and the intercept is far from 0.

````python
from hiring_cv_bias.reverse_matching.scoring import fit_weights

estimated = fit_weights(pairs_clean)
weights = {k: round(v, 1) for k, v in estimated.items()}
display(
    pl.DataFrame(
        [
            {"weights": "estimated, without outliers", **estimated},
            {"weights": "rounded to one decimal", **weights},
            {
                "weights": "estimated, all pairs",
                **fit_weights(pairs_all),
            },
        ]
    )
)
````

#### Exact check

A good fit is not enough: an approximate formula would leave a small error on every pair. So we count the scores that
the rounded formula reproduces exactly (`share_exact`, error below 10⁻⁴). On the pairs without outliers it is almost
all of them, so the score *is* this formula. The printed example recomputes one pair by hand.

The formula does not tell us where the outliers come from. It only tells us that they do not follow it: with every
component at its maximum of 1, the formula gives 2 + 1 + 1 + 1 + 3 × 0.3 = 5.9, below the outlier threshold. So the
outlier scores contain something outside the seven components, something the data does not show, and we leave those
pairs out of the study.

````python
from hiring_cv_bias.reverse_matching.scoring import validate_formula

display(validate_formula(pairs_clean, weights))

example = pairs_clean.row(0, named=True)
terms = " + ".join(f"{weights[c]:g}·{example[c]:.4f}" for c in COMPONENTS if example[c])
formula_value = sum(weights[c] * example[c] for c in COMPONENTS)
print(
    f"worked example, pair ({example['CANDIDATE_ID']}, {example['Job_id']}): "
    f"{terms} = {formula_value:.4f}; Adecco's score = {example['match_score']:.4f}"
)
````

#### Component rules

Five of the seven components can be recomputed from two things we have: the candidate's profile and the job's
requirements (the `Jobs` sheet). For a pair:

- `Card_match`: 1 if the candidate has every driving license the job asks for, else 0.
- `Languages_match`: the same with the languages.
- `Regulated_Trainings_match`: the same with the regulated trainings, the formal qualifications a job can require (in
  the example below, a field of study).
- `match_skills`: the share of the skills the job asks for that the candidate has (`matterknowledge` in the data).
- `Distance_match`: from the distance bands above.

For the three binary components, a job that asks for nothing gives 1; for `match_skills` it gives 0 (see below).

The table describes the job requirements behind these rules, one row per requirement (`req_licenses` for `Card_match`,
`req_languages` for `Languages_match`, `req_regulated` for `Regulated_Trainings_match`, `req_skills` for
`match_skills`). `share_jobs` is the share of jobs that ask for at least one item of that kind, and `example` is the
requirement of the first such job.

````python
from hiring_cv_bias.reverse_matching.components import job_requirements

jobs_req = job_requirements(raw["Jobs"])
display(
    jobs_req.select(
        pl.struct(
            share_jobs=(pl.col(c).list.len() > 0).mean(),
            example=pl.col(c).filter(pl.col(c).list.len() > 0).first().list.join("; "),
        ).alias(c)
        for c in ["req_licenses", "req_languages", "req_regulated", "req_skills"]
    )
    .unpivot(variable_name="requirement")
    .unnest("value")
)
````

To check the rules, we apply them to the real profile (P) of every pair and compare the result with the value Adecco
gives in `ReverseMatching.xlsx`. The **agreement** of a component is the share of pairs where the two values are
equal. All five agree on more than 99% of the pairs, so the rules are the ones Adecco uses.

`Interest` comes from the categories the candidate selects, not from the CV. The parser does not extract it, so it is
the same in every version of the dataset. `Experience` cannot be recomputed from the data; we estimate it in 1.3.

````python
from hiring_cv_bias.reverse_matching.audit import rules_agreement
from hiring_cv_bias.reverse_matching.profiles import build_profile_real

display(rules_agreement(pairs_all, build_profile_real(raw["Candidates"]), jobs_req))
````

`match_skills` needs one more choice: when a job asks for no skill, the share is 0/0. The first table takes the pairs
whose job has an empty `req_skills` and counts Adecco's `match_skills`. Almost all are 0, so our rule gives 0.

The second table lists the exceptions: pairs where Adecco gives a share of skills (1, 0.889, 0.5) although the `Jobs`
sheet lists no skill for the job. The values look like shares of a real list (0.889 = 8 of 9 skills), so Adecco
probably scored these jobs with a skill list that is not in our export. We cannot check it. These pairs are part of
the reason why the agreement of `match_skills` is not 1. In P, O and C our rule gives them 0.

````python
no_skills = pairs_all.join(jobs_req.select("Job_id", "req_skills"), on="Job_id").filter(
    pl.col("req_skills").list.len() == 0
)
display(no_skills.group_by("match_skills").len().sort("match_skills"))
display(
    no_skills.filter(pl.col("match_skills") > 0)
    .select("Job_id", "CANDIDATE_ID", "match_skills")
    .sort("Job_id")
)
````

### 1.3 Estimating `Experience` from job titles

To re-score O and C we need all seven components for each version. Six are recomputed with the rules of 1.2.
`Experience` cannot be. Adecco's documentation defines it as the "highest standardized similarity between the job
title and the candidate's experience", computed with an internal model we do not have; and the free-text job title is
not in the data either. We estimate it with the same rule, using the job category in place of the title. For each pair
the estimate is 1 if the job category appears verbatim among the candidate's titles, 0 if the candidate has no titles,
and otherwise the highest cosine similarity between the category and the titles (the past job closest to the offer).
The cosine comes from a sentence encoder, a model that turns a text into a vector, so that texts with a similar
meaning get similar vectors.

#### Choosing the encoder

Which encoder? On the real profile P we know Adecco's `Experience` for every pair, so we can try several encoders and
keep the one that reproduces it best. We compare five encoders and a string-matching baseline, all with the same rule
above, on the pairs without outliers.

The **lexical baseline** does not use any model: it only looks at the characters and the words of the two texts. It
computes two numbers and keeps the larger. The first is the share of characters the two strings have in common, in the
same order: `autista` and `autista di ambulanza` share the 7 characters of the first inside the 20 of the second, so 2
· 7 / (7 + 20) = 0.52. The second is the share of words in common: `autista` is one of the three distinct words, 0.33.
The baseline gives 0.52. It is the simplest possible estimator, and it tells us whether an encoder adds anything over
plain string matching.

**Why multilingual encoders.** The texts are mostly Italian, but not only. The parser's job titles follow an English
taxonomy and some were never translated, and the recruiters write part of the job categories in English (`sales
representative`, `cost engineer`) or in both languages (`commesso di punto vendita/shop assistant`). The table below
gives the share of texts that a language detector (Italian or English only) reads as English, for the titles of P
(counting every occurrence) and for the job categories (one per job). An Italian-only encoder cannot see that
`warehouse worker` and `magazziniere` are the same job, so we include an Italian-only encoder in the comparison, to
measure what that costs.

````python
from hiring_cv_bias.reverse_matching.experience import language_shares

pairs_P = pairs_clean.join(clean.select("CANDIDATE_ID"), on="CANDIDATE_ID")
profile_P = build_profile_real(raw["Candidates"])
titles_P = profile_P.explode("experience").drop_nulls("experience")["experience"]
display(
    pl.concat(
        [
            language_shares(titles_P.to_list()).with_columns(
                pl.lit("titles of P").alias("texts")
            ),
            language_shares(jobs_req["job_category"].to_list()).with_columns(
                pl.lit("job categories").alias("texts")
            ),
        ]
    ).pivot(on="language", index="texts", values="share")
)
````

A good minority of the titles and a few of the job categories are English; the mixed ones count as Italian here.

The encoders are: `MiniLM-L12` and `mpnet-base` (two multilingual sentence-transformers models trained on
paraphrases), `e5-base` and `e5-large` (Microsoft's multilingual E5 family, trained for retrieval), and the
Italian-only `sentence-BERTino` (an Italian BERT fine-tuned on translated sentence-similarity data). The
`parameters_M` column gives their size in millions of parameters. In the multilingual models a large part of these
parameters is the vocabulary table of the word pieces of many languages, so the size says little about the quality on
short Italian texts. The comparison is fair because all models get the same texts and the same rule, and the result on
P, not the size, decides.

The other columns:

* `pearson` and `spearman` – correlation of the values and of the rankings with Adecco's `Experience`. The ranking is
  what matters for a position in a job's list.
* `spearman_non_verbatim` – the same ranking correlation, restricted to the pairs with titles but no verbatim match:
  the only pairs where the estimator actually decides something.
* `mae` – mean absolute error. `Experience` weighs 2 in the score, so an error of e moves the score by 2e.
* `min_with_titles` – the lowest estimate among the pairs whose candidate has titles. The range of Adecco's value on
  these pairs is printed above the table: an encoder on the right scale should not go below it.
* `n_far_below` – pairs estimated below 0.4 while Adecco gives more than 0.8: gross ranking errors that no rescaling
  can fix.

The first run downloads the five models from Hugging Face (about 5 GB). The embeddings are then cached in
`data/ethics/embeddings/`, and a model whose embeddings are all cached is not downloaded.

````python
from hiring_cv_bias.reverse_matching.build import EMBEDDINGS_DIR
from hiring_cv_bias.reverse_matching.experience import (
    candidate_estimators,
    compare_estimators,
    probe_table,
)

estimators = candidate_estimators(EMBEDDINGS_DIR)
with_titles = pairs_P.filter(pl.col("Experience") > 0)["Experience"]
print(
    f"Adecco's Experience on the {with_titles.len()} pairs with titles: "
    f"min {with_titles.min():.3f}, max {with_titles.max():.3f}"
)
display(compare_estimators(pairs_P, profile_P, jobs_req, estimators))
````

The encoders differ a lot, and size does not explain the differences. `MiniLM-L12`, the smallest multilingual model,
ranks the pairs worse than the string baseline and makes hundreds of gross errors. `mpnet-base` and `e5-base` have the
same architecture and the same size, but the first still puts many pairs far below Adecco's value and the second none:
the difference is the training, not the size. The largest model is not better than `e5-base`: `e5-large`, from the
same family, is a little worse on every column.

The Italian-only model confirms the choice of a multilingual encoder. `sentence-BERTino` ranks the pairs as well as
`e5-base`, but it makes many gross errors, its lowest estimate is far below Adecco's range and its mean absolute error
is ten times larger. So it would need a calibration step.

`e5-base` has no gross error and the highest correlations, also on the non-verbatim pairs. And, unlike all the others,
it is on Adecco's scale: its lowest estimate on a pair with titles is just below Adecco's lowest value, and its mean
absolute error is a few hundredths.

The table below shows where the other encoders fail. The first five rows are short Italian titles against a longer
category that contains them. The sub-word tokenizer of MiniLM splits `autista` into `aut` + `ista` and `commesso` into
`comme` + `sso`, and the vector of the single word ends up far from the compound title. These are exactly the pairs
that Adecco rates around 0.9. The last two rows cross the languages.

````python
display(probe_table(estimators))
````

On the short Italian titles, `MiniLM-L12` gives a fraction of what `e5-base` gives, and even the string baseline does
better than MiniLM there. Across languages the Italian-only model collapses: on the two cross-language rows,
`sentence-BERTino` is far below every multilingual encoder, while `e5-base` rates them like the Italian pairs.

We therefore use **`multilingual-e5-base`**. Its cosine is already on Adecco's scale, so the estimate is the cosine
itself, with no calibration step: a monotone rescaling learned on P would add a fitted component to the pipeline
without changing any ranking.

The same estimator is used for P, O and C. P also uses the *estimated* `Experience`, so the three versions are scored
in exactly the same way, and differences between versions come only from the titles.

````python
from hiring_cv_bias.reverse_matching.audit import profile_field_stats

profiles = result.profiles
display(profile_field_stats(profiles))
````

#### The real profile is not the parser output

If P came from the parser, it would look like O. It does not: P has more licenses and fewer languages than O. The
table compares P and O candidate by candidate. The parser found a license for about two thousand candidates, but only
a small minority of them have any license in P. It found languages for most candidates, but less than half of them
have a language in P, and when both have one, the lists rarely share an item. Job titles are different: every
candidate with parsed titles has titles in P, and half of them share at least one. So `experience` comes from the
parser (titles translated to Italian, plus titles added by hand: the lists are almost never identical), while licenses
and languages were entered by hand. The real matcher used a **partly manual** profile.

We do not know where the manual fields come from: a correction of the parser, or information that was never in the CV
(for example, from an interview). So we do not use P to judge the parser: the parser's bias is measured on the
Original vs Corrected comparison only. P serves to validate the rules (1.2) and the `Experience` estimate (1.3).

**How to read the table.** For each field, we look only at the candidates where the parser found something. Then we
ask three questions, each one about a smaller group than the one before:

1. **Does the real profile have anything in this field?** (`share_also_in_real`)
2. **If yes, do the two lists share at least one item?** (`share_common_item`)
3. **Are the two lists exactly the same?** (`share_identical`)

`candidates_parser` is the number of candidates we start from.

*Example.* The parser finds the skills `[Excel, SQL]` for a candidate:
- real profile `[]` → counts as "no" for question 1, and is left out of questions 2 and 3;
- real profile `[SQL, Python]` → yes to question 2, no to question 3;
- real profile `[SQL, Excel]` → yes to both (order does not matter).

````python
from hiring_cv_bias.reverse_matching.audit import profile_overlap

display(profile_overlap(profiles["P"], profiles["O"]))
````

#### Differences at the pair level

The table above compares the profiles. Here we look at what those differences do to the score. For every (candidate,
job) pair we compute the components and the score with each version, then we count how often the value changes. Each
column compares two versions:

- **P vs O**: the real profile against the parser output;
- **O vs C**: the parser output against the corrected profile.

A cell is the share of pairs where the value is different. For `Experience_hat`, a gap of 0.01 or less counts as no
change.

What the table shows:

- `Regulated_Trainings_match`, `match_skills` and `Distance_match` never change. They do not depend on what the parser
  extracts: regulated trainings and skills are copied from P, and the distance does not use the profile.
- **P vs O.** `Card_match` and `Languages_match` differ in a minority of the pairs, but `Experience_hat` and the score
  differ in almost all of them, because the parser's job titles are almost never the same as the real ones
  (`share_identical` in the table above).
- **O vs C.** Correcting the parser changes the two binary components in less than one pair in ten, `Experience_hat`
  in about one in six, and the score in about one in three.

````python
from hiring_cv_bias.reverse_matching.audit import version_differences

display(
    pl.concat(
        [version_differences(pairs, "P", "O"), version_differences(pairs, "O", "C")]
    ).pivot(on="comparison", index="column", values="share_different")
)
````
